```python
import requests, getpass

TOKEN = getpass.getpass("Enter bot id and press Enter: ").strip()

r = requests.get(f"https://api.telegram.org/bot{TOKEN}/getUpdates").json()
print("Telegram:", r.get("ok"), r.get("description", ""))

ids = {}
for u in r.get("result", []):
    m = u.get("message")
    if m:
        ids[m["chat"]["id"]] = m["chat"].get("first_name", "")
print("Mili IDs:", ids)

for cid in ids:
    resp = requests.post(f"https://api.telegram.org/bot{TOKEN}/sendMessage",
                         data={"chat_id": cid, "text": "Test alert from Truck.AI"}).json()
    print(cid, "->", "sended" if resp.get("ok") else resp.get("description"))
```
